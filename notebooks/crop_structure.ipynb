```python
import pandas as pd
import numpy as np
```

```python
df_list = pd.read_html("horizontal_crop_vertical_year_report_maharashtra.xls")
df = df_list[0]

df.head()
```

```python
df.columns.tolist()
```

```python
df.columns = [
    f"{c1}_{c2}_{c3}" if c1 not in ['State', 'District', 'Year'] else c1
    for (c1, c2, c3) in df.columns
]

df.head()
```

```python
df.columns = df.columns.str.replace(" ", "_")
df.columns = df.columns.str.replace("/", "_")
df.columns = df.columns.str.replace("(", "")
df.columns = df.columns.str.replace(")", "")

df.columns.tolist()
```

```python
cols = [col for col in df.columns if col not in ['State', 'District', 'Year']]
```

```python
rows = []

for col in cols:
    parts = col.split("_")
    
    # Extract components
    crop = parts[0]
    if len(parts) > 3:
        crop = "_".join(parts[:-3])  # handles names like "Other_Summer_Pulses"
    
    season = parts[-3]
    metric = parts[-2]  # Area / Production / Yield
    
    for i in range(len(df)):
        rows.append({
            'State': df.loc[i, 'State'],
            'District': df.loc[i, 'District'],
            'Year': df.loc[i, 'Year'],
            'Crop': crop,
            'Season': season,
            'Metric': metric,
            'Value': df.loc[i, col]
        })

long_df = pd.DataFrame(rows)
long_df.head()
```

```python
master_df = long_df.pivot_table(
    index=['State', 'District', 'Year', 'Crop', 'Season'],
    columns='Metric',
    values='Value'
).reset_index()

master_df.head()
```

```python
master_df.columns.name = None

master_df = master_df.rename(columns={
    'Area': 'Area',
    'Production': 'Production',
    'Yield': 'Yield'
})
```

```python
master_df['Year'] = master_df['Year'].astype(str).str[:4].astype(int)
```

```python
master_df['Yield'] = master_df['Production'] / master_df['Area']
```

```python
master_df.head()
```

```python
master_df = master_df.replace([np.inf, -np.inf], np.nan)
master_df = master_df.dropna(subset=['Yield'])
master_df = master_df[master_df['Area'] > 0]
```

```python
master_df = master_df.drop(columns=['State','Bales', 'Tonne'], errors='ignore')
```

```python
master_df.head()
master_df.columns
```

```python
master_df.groupby('District')['Year'].agg(['min', 'max', 'nunique'])
```

```python
master_df = master_df.sort_values(['District', 'Crop', 'Season', 'Year'])
```

```python
master_df['Yield_Lag1'] = (
    master_df.groupby(['District', 'Crop', 'Season'])['Yield']
    .shift(1)
)
```

```python
master_df['Yield_Lag2'] = master_df.groupby(['District','Crop','Season'])['Yield'].shift(2)
```

```python
master_df['Yield_3yr_Avg'] = (
    master_df.groupby(['District', 'Crop', 'Season'])['Yield']
    .transform(lambda x: x.rolling(3).mean())
)
```

```python
master_df['Yield_3yr_Std'] = (
    master_df.groupby(['District', 'Crop', 'Season'])['Yield']
    .transform(lambda x: x.rolling(3).std())
)
```

```python
master_df.head()
```

```python
# Keep lag1 but fill missing with current yield
master_df['Yield_Lag1'] = master_df['Yield_Lag1'].fillna(master_df['Yield'])
```

```python
master_df['Yield_Lag2'] = master_df['Yield_Lag2'].fillna(master_df['Yield_Lag1'])
```

```python
master_df['Yield_3yr_Avg'] = master_df['Yield_3yr_Avg'].fillna(master_df['Yield'])
```

```python
master_df['Yield_3yr_Std'] = master_df['Yield_3yr_Std'].fillna(0)
```

```python
master_df.isna().sum()
```

```python
cols = [
    'District', 'Year', 'Crop', 'Season',
    'Area', 'Production','Yield',
    'Yield_Lag1', 'Yield_Lag2',
    'Yield_3yr_Avg', 'Yield_3yr_Std'
]

master_df = master_df[cols]
```

```python
master_df.head()
```

```python
master_df['District'].unique()
```

```python
master_df['District_Num'] = master_df['District'].str.extract(r'(\d+)').astype(int)
```

```python
master_df['District_Name'] = master_df['District'].str.replace(r'^\d+\.\s*', '', regex=True)
```

```python
master_df = master_df.sort_values(['District_Num', 'Year'])
```

```python
master_df['District'] = master_df['District_Name']
master_df = master_df.drop(columns=['District_Name', 'District_Num'])
```

```python
master_df.to_csv("master_crop_data_clean.csv", index=False)
```
