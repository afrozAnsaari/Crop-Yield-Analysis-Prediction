```python
import pandas as pd
```

```python
df_rain=pd.read_csv("data/maharain_district_rainfall_1998_2022.csv")
df_crop=pd.read_csv("data/master_crop_data_clean.csv")
```

```python
# Makes sure bothe the columns(district and location name) have same name cases i.e., lower
def clean_text(col):
    return col.str.lower().str.strip()

df_crop["District"] = clean_text(df_crop["District"])
df_rain["location_name"] = clean_text(df_rain["location_name"])
```

```python
df_crop["District"].unique()
```

```python
df_rain["location_name"].unique()
```

```python
set(df_crop["District"]) - set(df_rain["location_name"])
```

```python
# THIS ERROR OCCURED AS BOTH THE DATASET HAVE NAME MISMATCH:
#"raigad": "raigadh",
#"parbhani": "parabhani"
```

```python
# NAME MISMATCH FIX
mapping_fix = {
    "raigadh": "raigad",
    "parabhani": "parbhani"
}

df_rain["location_name"] = df_rain["location_name"].replace(mapping_fix)
```

```python
df_rain["location_name"].unique()
```

```python
df_crop["District"].unique()
```

```python
set(df_crop["District"]) - set(df_rain["location_name"])
```

```python
# NOW THE NAME MISMATCH HAS BEEN RESOLVED AND ONLY THE MISSING DATASET OF MUMBAI AND MUM. SUBS ARE IN DISJOINT
```

```python
df=pd.merge(
    df_crop,
    df_rain, 
    left_on=["Year", "District"],
    right_on=["year", "location_name"],
    how="inner"
)
```

```python
df = df.drop(columns=[
    "year",
    "location_name",
    "division_code",
    "division_name",
    "row_order"
])
```

```python
import os

folder = "data"
filename = "final_merged_data_rain_plus_crop_yield.csv"
path = os.path.join(folder, filename)

# create folder if it doesn't exist
os.makedirs(folder, exist_ok=True)

if not os.path.exists(path):
    df.to_csv(path, index=False)
    print(f"File saved{path}")
else:
    print(f"File already exists at path: {path}, skipping...end")

```
